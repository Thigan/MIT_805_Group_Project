```python
import os
import shutil
```

```python
WORKING_DATA_FOLDER = r"C:\Users\thiga\OneDrive\Documents\Masters - Big Data\Year 1\Semester 2\MIT 805\nyc_tlc_working"
SAMPLE_FOLDER = r"C:\Users\thiga\OneDrive\Documents\Masters - Big Data\Year 1\Semester 2\MIT 805\nyc_tlc_working_sample"

TARGET_SIZE_GB = 5
os.makedirs(SAMPLE_FOLDER, exist_ok=True)
```

### Get a list of all the parquet files in date order

```python
all_file_names = []
for file_name in os.listdir(WORKING_DATA_FOLDER):
    if file_name.endswith(".parquet"):
        all_file_names.append(file_name)

all_file_names = sorted(all_file_names)
print("Total files found: " + str(len(all_file_names)))
```

### Add up the size of every file so we know the size of full working dataset

```python
total_size_bytes = 0
for file_name in all_file_names:
    file_path = os.path.join(WORKING_DATA_FOLDER, file_name)
    total_size_bytes = total_size_bytes + os.path.getsize(file_path)

total_size_gb = total_size_bytes / (1024 ** 3)
print("Total size of all files: " + str(round(total_size_gb, 2)) + " GB")
```

### Work out how many files needed to achive Target_Size_GB

```python
fraction_to_keep = TARGET_SIZE_GB / total_size_gb
number_of_files_to_keep = round(len(all_file_names) * fraction_to_keep)

if number_of_files_to_keep < 1:
    number_of_files_to_keep = 1

print("Number of files we will keep: " + str(number_of_files_to_keep))
```

### Pick files evenly across the date range to include different years and seasons.

```python
gap_between_files = len(all_file_names) / number_of_files_to_keep

selected_file_names = []
for i in range(number_of_files_to_keep):
    index = int(i * gap_between_files)
    selected_file_names.append(all_file_names[index])
```

### Copy the selected files into the new sample folder

```python
running_total_bytes = 0

for file_name in selected_file_names:
    source_path = os.path.join(WORKING_DATA_FOLDER, file_name)
    destination_path = os.path.join(SAMPLE_FOLDER, file_name)

    shutil.copy2(source_path, destination_path)

    running_total_bytes = running_total_bytes + os.path.getsize(source_path)
    running_total_gb = running_total_bytes / (1024 ** 3)
    print("Copied " + file_name + " - running total: " + str(round(running_total_gb, 2)) + " GB")


print("Done")
print("Kept " + str(len(selected_file_names)) + " files")
print("Total sample size: " + str(round(running_total_bytes / (1024 ** 3), 2)) + " GB")
print("Saved to: " + SAMPLE_FOLDER)
```
